# 01 — MOE Feature Builder

This notebook runs the full Stage 1 pipeline:

1. Validate all raw datasets
2. Stop if validation fails
3. Build the canonical daily feature dataset
4. Run feature quality checks
5. Save all outputs


````python
#=========================================================
# project setup
#=========================================================

from pathlib import Path
import sys

project_root = Path.cwd()

if project_root.name == "notebooks":
    project_root = project_root.parent

src_path = project_root / "src"

if str(src_path) not in sys.path:
    sys.path.insert(0, str(src_path))

print(f"project root: {project_root}")
````

````python
#=========================================================
# imports
#=========================================================

from validate_data import (
    validate_raw_folder,
    assert_validation_passed,
)

from build_features import build_features

from quality_checks import (
    create_quality_outputs,
    assert_feature_quality_passed,
)
````

````python
#=========================================================
# step 1 - validate raw datasets
#=========================================================

validation_report = validate_raw_folder()

display(validation_report)

assert_validation_passed(
    validation_report
)

print("\nraw data validation passed")
````

````python
#=========================================================
# step 2 - build canonical feature dataset
#=========================================================

features = build_features()

print("\n=====================================")
print("==feature dataset")
print(f"rows: {len(features):,}")
print(f"columns: {len(features.columns):,}")
print(f"date min: {features['date_day'].min().date()}")
print(f"date max: {features['date_day'].max().date()}")

display(features.head())
````

````python
#=========================================================
# step 3 - run feature quality checks
#=========================================================

quality_report, feature_summary = create_quality_outputs(
    feature_df=features
)

display(quality_report)

assert_feature_quality_passed(
    quality_report
)

print("\nfeature quality checks passed")
````

````python
#=========================================================
# inspect feature summary
#=========================================================

display(
    feature_summary.sort_values(
        [
            "null_percentage",
            "column",
        ],
        ascending=[
            False,
            True,
        ],
    )
)
````
